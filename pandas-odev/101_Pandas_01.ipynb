## Bölüm 2

* `mount_everest_deaths` veri kümesinin kendi indeks sütunu vardır. Veriyi içe aktarırken bu mevcut indeks sütununu kullanın.  
* Hangi sütunlarda sıfır eksik değer (null) vardır?  
* Hangi sütunda en fazla eksik değer (null) var?


```python
everest = pd.read_csv("mount_everest_deaths.csv", index_col=0)
nulls = everest.isnull().sum()
print(nulls)
print("Eksik değeri olmayan sütunlar:", list(nulls[nulls == 0].index))
print("En fazla eksik değere sahip sütun:", nulls.idxmax())
```
