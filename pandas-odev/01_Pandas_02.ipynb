**Aşağıdaki soruları yanıtlamak için `bestsellers.csv` veri kümesi ile çalışın:**

```python
import pandas as pd

bestsellers = pd.read_csv("/Users/d3-13/Desktop/WS/Odve/bestsellers.csv")
```

## Bölüm 1

* Kitap adlarını içeren bir pandas Series elde edin.  
* Kullanıcı puanlarını içeren bir Series elde edin.  
* İlk 8 yazarı alın.


```python
names = bestsellers["Name"]
print(names)

ratings = bestsellers["User Rating"]
print(ratings)

print(bestsellers["Author"].head(8))
```

## Bölüm 2

* Eşsiz (unique) Türleri (Genre) bulun.  
* Kaç farklı yazar olduğunu bulun.  
* Ortalama fiyatı bulun.  
* En yüksek 10 fiyatı bulun.


```python
print("Eşsiz türler:", bestsellers["Genre"].unique())
print("Farklı yazar sayısı:", bestsellers["Author"].nunique())
print("Ortalama fiyat:", bestsellers["Price"].mean())
print("En yüksek 10 fiyat:")
print(bestsellers["Price"].nlargest(10))
```

## Bölüm 3

* Veri kümesindeki en yaygın 3 kitap adını bulun.  
* Sadece Yazar ve Kullanıcı Puanı içeren yeni bir DataFrame oluşturun.  
* Yeni DataFrame'i kullanarak en yaygın Yazar ve Kullanıcı Puanı kombinasyonunu bulun.


```python
print(bestsellers["Name"].value_counts().head(3))

author_rating = bestsellers[["Author", "User Rating"]]
print(author_rating.head())

print(author_rating.value_counts().head(1))
```

## Bölüm 4

* Veri kümesinde her bir Tür'ün (Genre) kaç kez geçtiğini gösteren bir pasta (pie) grafiği oluşturun.  
* En üretken 10 yazarı bulun ve kitap sayılarını çubuk grafik (bar plot) olarak görselleştirin.  
* Kullanıcı puanlarının (rating) dağılımını gösteren bir histogram grafiği oluşturun.


```python
import matplotlib.pyplot as plt

bestsellers["Genre"].value_counts().plot(kind="pie", autopct="%1.1f%%", title="Türlere Göre Dağılım")
plt.ylabel("")
plt.show()

bestsellers["Author"].value_counts().head(10).plot(kind="bar", title="En Üretken 10 Yazar")
plt.ylabel("Kitap Sayısı")
plt.show()

bestsellers["User Rating"].plot(kind="hist", bins=20, title="Kullanıcı Puanı Dağılımı")
plt.xlabel("User Rating")
plt.show()
```
